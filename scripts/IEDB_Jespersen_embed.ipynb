```python
import sys
from pathlib import Path
PROJECT_ROOT = Path.cwd().parents[1]   # move up from notebooks/
sys.path.insert(0, str(PROJECT_ROOT))
from data.pre_embed import SingleSequenceDS
import torch
from data.embed import ESMCBatchEmbedder
data_name = 'IEDB_Jespersen'
model_name = 'esmc_300m'
base_data_dir = Path.cwd().parent / 'data' / 'data_files'
```

```python
from data.parse import get_tdc_epitope
raw_data = get_tdc_epitope(data_name, file_dir=base_data_dir)
ds_saver = SingleSequenceDS(data_name, df=raw_data, save_dir=base_data_dir)
```

Embed Data via ESMC
If above cell run locally, upload data_files to colab

```python
sequence_dataset = SingleSequenceDS(data_name, save_dir=base_data_dir)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
embedder = ESMCBatchEmbedder(model_name, save_dir=base_data_dir / data_name, device=device)
sequences_to_embed = sequence_dataset.unique_sequences
embedder.batch_save(sequences_to_embed, max_tok_per_batch=10000)
```
